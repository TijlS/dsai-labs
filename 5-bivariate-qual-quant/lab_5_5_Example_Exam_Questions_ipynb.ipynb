```python
# Package imports
import numpy as np                                  # "Scientific computing"
import scipy.stats as stats                         # Statistical tests

import pandas as pd                                 # Dataframe
import matplotlib.pyplot as plt                     # Basic visualisation
from statsmodels.graphics.mosaicplot import mosaic  # Mosaic plot
import seaborn as sns                               # Advanced dataviz
```

## Question 1

The data frame below shows the evolution of the average duration of a first marriage in Belgium, since 2015 by type of marriage (male-female / female-female / male-male) (Source: StatBel). Can we say that a man-man marriage lasts significantly longer than a woman-woman marriage?  
Make a plot!



```python
divorces = pd.DataFrame(data={
    'year': [2015,2016,2017,2018,2019,2020,2021,2022,2023],
    'm_f': [16.7, 16.9, 16.9, 16.7, 16.7, 16.4, 16.6, 16.7, 16.1],
    'f_f': [5.5, 6.4, 6.3, 7.4, 6.7, 6.7, 7.0, 7.8, 8.0],
    'm_m': [6.0, 6.3, 6.9, 7.4, 7.4, 8.0, 8.2, 8.2, 8.5]
}).set_index('year')
divorces.head()
```

```python
sns.boxplot(data=divorces[['m_m', 'f_f']], orient='h')
```

```python
# Stap 1 - hypothese
# H₀ => Geen significant verschil in de lengte van de huwelijken
# H₁ => man-man duren langer dan vrouw-vrouw

# Stap 2 - significance level
alpha = 0.05

# Stap 3 - teststatistiek
# Stap 4 - p value
t, p = stats.ttest_ind(divorces['m_m'], divorces['f_f'], alternative='greater', equal_var=False)
print(f't-value  : {t}')
print(f'p-value  : {p}')
# Stap 5 - Conclusie
# p > geen significant verschil in lengte tussen de huwelijken
```

## Question 2

The data frame below shows the evolution of the average duration of a first and a subsequent marriage in Belgium since 2015 (Source: StatBel). Can we say that a first marriage lasts significantly longer than a subsequent marriage? Make a plot!


```python
divorces = pd.DataFrame(data={
    'year': [2015,2016,2017,2018,2019,2020,2021,2022,2023],
    'first_marriage': [16.7, 16.9, 16.9, 16.7, 16.7, 16.4, 16.6, 16.7, 16.1],
    'subsequent_marriage': [10.5, 11.0, 11.2, 11.3, 11.3, 11.4, 11.5, 11.9, 11.8]
}).set_index('year')
divorces.head()
```

```python
sns.boxplot(
    data=divorces[['first_marriage', 'subsequent_marriage']],
    orient='h'
)
```

```python
# Stap 1 - hypothese
# H₀ => Geen significant verschil in de lengte van de huwelijken
# H₁ => Eerste huwelijk duurt langer dan 2e

# Stap 2 - significance level
alpha = 0.05

# Stap 3 - teststatistiek
# Stap 4 - p value
t, p = stats.ttest_ind(divorces['first_marriage'], divorces['subsequent_marriage'], alternative='greater', equal_var=False)
print(f't-value  : {t}')
print(f'p-value  : {p}')
# Stap 5 - Conclusie
# p << H0 verworpen, eerste huwelijk duurt significant langer dan volgende
```

## Question 3
A local pizzeria claims that they can deliver pizza faster than a local branch of a large chain. To verify this, you ordered pizza from both of them 10 times. Below are the times (in minutes) they took. Can the local pizzeria indeed deliver pizzas significantly faster than the big player? Make a plot.



```python
dfpizzas = pd.DataFrame(data= {'local': [16.8, 11.7, 15.6, 16.7, 17.5, 18.1, 14.1, 21.8, 13.9, 20.8],
                               'chain': [22.0, 15.2, 18.7, 15.6, 20.8, 19.5, 17.0, 19.5, 16.5, 24.0]})
dfpizzas.head()
```

```python
sns.boxplot(
    data=dfpizzas[['local', 'chain']],
    orient='h'
)
```

```python
# Stap 1 - hypothese
# H₀ => Geen significant verschil in leveringstijd lokale italiaan en keten
# H₁ => Lokale italiaan is sneller (Luigi op z'n vespa)

# Stap 2 - significance level
alpha = 0.05

# Stap 3 - teststatistiek
# Stap 4 - p value
t, p = stats.ttest_ind(dfpizzas['local'], dfpizzas['chain'], alternative='less', equal_var=False)
print(f't-value  : {t}')
print(f'p-value  : {p}')
# Stap 5 - Conclusie
# p > alpha => niet voldoende reden om H0 te verwerpen
```
